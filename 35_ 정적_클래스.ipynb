```python
# 클래스 속성 사용하기

# class 클래스이름:
#     속성 = 값

class Person:
    bag = []

    def put_bag(self, stuff):
        Person.bag.append(stuff)


james = Person()

james.put_bag("책")

maria = Person()

maria.put_bag("열쇠")




print(james.bag)
print(maria.bag)
```

```python
print(Person.bag)
```

```python
# 인스턴스 속성 사용하기
class Person:
    bag = []

    def __init__(self):
        pass


    def put_bag(self, stuff):
        Person.bag.append(stuff)


james = Person()

james.put_bag("책")

maria = Person()

maria.put_bag("열쇠")




print(james.bag)
print(maria.bag)
```

```python
# 비공개 클래스 속성 사용하기
## class 클래스이름:
##     __속성=값

class knight:
    __item_limit = 10
        
    def print_item_limit(self):
        print(knight.__item_limit)


x = knight()
x.print_item_limit()


print(knight.__item_limit())
```

```python
# 정적매서드 사용하기

# class 클래스이름:
#     @staticmethod
#     def 매서드(매개변수):
#         코드



class Calc:
    @staticmethod
    def add(a,b):
        return a + b
    
    @staticmethod
    def mul(a,b):
        return a * b


print(Calc.add(10, 20))
print(Calc.mul(10, 20))

```

```python
#클래스 매서드 사용하기

# class 클래스이름:
#     @classmethod
#     def 매서드(cls, 매개변수1, 매개변수2):
#         코드



class Person:
    count = 0

    def __init__(self):
        Person.count +=1

    @classmethod
    def print_count(cls):
        print(f"{cls.count}명 생성 되었습니다.")

james = Person()
maira = Person()

Person.print_count()
```
