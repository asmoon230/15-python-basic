```python
# 딕셔너리 표현식 사용하기 (딕셔너리 컴프리헨션)

## 1. {키 : 값 for 키, 값 in 딕셔너리}
## 2. dict({키 : 값 for 키, 값 in 딕셔너리})

keys = ["a", "b", "c","d"]
o = [["a",1],["b",2],["c",3]]
k = dict.fromkeys(keys).items()
x = {key:value for key , value in o}
print(k)
print(x)
```

```python
# 딕셔너리 표현식 에서 if 조건문 사용하기

x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
} 

# for key, value in x.items(): 읽어와야하는 원본값을 삭제하여서 런타임에러 (딕셔너리는 인덱스값으로 반복문을 시행하는것이 아닌 각 key값에 대한 시행이라 key가 사라지면 에러)
for key, value in list(x.items()):
    if value ==20:
        del x[key]

print(x)

print(list(x.items()))
print(x.items())
```
