```python
# 집합연산 사용하기

## 세트1  세트2
## set.union(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a|b)
print(set.union(a,b))
print(a.union(b))
```

```python
# 교집합
## 세트1 & 세트2
## set.intersection(세트1, 세트2)


a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)
print(set.intersection(a,b))
print(a.intersection(b))
```

```python
# 차집합

## 세트1 - 세트2
## set.difference(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a,b))
print(a.difference(b))


print("="*10)


print(b - a)
print(set.difference(b,a))
print(b.difference(a))
```

```python
# 대칭 차집합

## 세트1 ^ 세트2
## set.symmetirc_difference(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a ^ b)
print(set.symmetric_difference(a,b))
print(a.symmetric_difference(b))

```

```python
# 집합 연산 후 할당 연산자 사용하기

## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
b = {3, 5, 6, 7}

x = {"apple","orange","peach"}
y = {"orange","peach","apple"}

a |= {5}

print(a)


a.update({6})
print(a)


c = dict(zip(a,b))
print(c)
print(type(c))

z = dict(zip(x,y))
print(z)
print(type(z))
#
#
#
#
#
#
#
#
```

```python
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}

print(a)

a = {1, 2, 3, 4}
a.intersection_update({0,1,2,3,4})
print(a)

```

```python
## 세트1 ^= 세트2
## 세트1.symmetric_difference_update(세트2)

a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)

a = {1, 2, 3, 4}
a.symmetric_difference_update({3, 4, 5, 6})


print(a)
```

```python
# 부분집합과 상위집합 확인하기

## 현재세트 <= 다른세트
## 현재세트.issubset(다른세트)

a = {1, 2, 3, 4}
print(a <={1,2,3,4,5}) 
print(a.issubset({1,2,3,4,5}) )

```

```python
## 현재세트 < 다른세트

a = {1, 2, 3, 4}
print(a <{1,2,3,4,5}) 


```

```python
## 현재세트 >= 다른세트
## 현재세트.issuperset(다른세트)


a = {1, 2, 3, 4}
print(a >={1,2,3,4,5}) 
print(a.issuperset({1,2,3,4,5}))
```

```python
## 현재세트 > 다른세트

a = {1, 2, 3, 4}
print(a > {1,2,3,4,5}) 

```
