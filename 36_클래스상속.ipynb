```python
# 사람클래스로 학생클래스 만들기

# class 기반클래스:
#     pass

# class 파생클래스(기반클래스):
#     pass


class Person:
    def greeting(self):
        print("안녕하세요")

class Student(Person):
    def study(self):
        print("공부하기")


james = Student()
james.greeting()
james.study()

```

```python
print(issubclass(Student, Person))
print(issubclass(Person, Student))
```

```python
# 기반클래스의 속성 사용하기

class Person:
    def __init__(self):
        print("Person__init___")
        self.hello = "안녕하세요"

class Student(Person):
    def __init__(self):
        print("Student__init___")
        self.school = "파이썬 코딩 도장"



james = Student()



print(james.school)
print(james.hello)
```

```python
# super()로 기반 클래스 초기화 하기
class Person:
    def __init__(self):
        print("Person__init___")
        self.hello = "안녕하세요"

class Student(Person):
    def __init__(self):
        print("Student__init___")
        super().__init__()
        self.school = "파이썬 코딩 도장"
```

```python
james = Student()



print(james.school)
print(james.hello)
```

```python
# 다중상속 사용하기

# class 기반클래스1:
#     코드

# class 기반클래스2:
#     코드

# class 파생클래스(기반클래스1, 기반클래스2):
#     코드

class Person:
    def greeting(self):
        print("안녕하세요")

class University:
    def menage_credit(self):
        print("학생관리")

class Undergraduate(Person, University):
    def study(self):
        print("공부하기")




james = Undergraduate()
james.greeting()
james.menage_credit()
james.study()
```

```python
# 추상 클래스 사용하기

# from abc import *

# class 추상클래스이름(metaclass = ABCmeta):
#     @abctractmethod
#     def 매서드이름(self):
#         코드


from abc import * 

class StudentBase(metaclass=ABCMeta):
    @abstractmethod
    def study(Self):
        pass

    @abstractmethod
    def go_to_school(self):
        pass

class Student(StudentBase):
    def study(self):
        print("공부하기")

james = Student()
james.study()
```

```python
from abc import * 

class StudentBase(metaclass=ABCMeta):
    @abstractmethod
    def study(Self):
        pass

    @abstractmethod
    def go_to_school(self):
        pass

class Student(StudentBase):
    def study(self):
        print("공부하기")
    def go_to_school(self):
        print("학교가기")


james = Student()
james.study()
```
