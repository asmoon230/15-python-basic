```python
# 딕셔너리 만들기

lux = {
    "health" : 490,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72 
}
print(type(lux))
print(lux)
```

```python
# 키 이름이 중복이 되면

lux = {
    "health" : 490,
    "health" : 800,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72 
}

print(lux) #컴퓨터는 위에서 아래로 왼쪽에서 오른쪽으로 읽기때문에 마지막 값을 읽는다
print(lux["health"])
```

```python
# 딕셔너리 키의 자료형
## 1. 문자형 가능
## 2. 정수형 가능
## 3. 실수형 가능
## 4. 불(True, False) 가능

## 불가능한 키의 자료형
## 1. 딕셔너리 불가능
## 2. 리스트 불가능
x = {
    100 : "hundred",
    False : 0,
    3.5 : [3.5, 3.5]
}
print(type(x))
print(x)
```

```python
y = {
    [10, 20] : 100
}
print(y)
print(type(y))
```

```python
z = {
    {"a" : 10} : 100
}

print(z)
print(type(z))
```

```python
f = {
    (10, 20) : 100
}

print(f)
print(type(f))
```

```python
#빈딕셔너리 만들기

x = {}
y = dict()

print(x)
print(type(x))
print(bool(x))
print(type(y))
print(y)
print(bool(y))
```

```python
# dict로 딕셔너리 만들기

lux1 = dict(health=400, mana=334, melee=550, armor=19.72)

print(lux1)
print(type(lux1))
```

```python
lux2 = zip(["health", "mana", "melee", "armor"], [490, 334, 550, 18.72]) #zip 안에 리스트를 넣어서 결합
print(lux2)
print(type(lux2))
#질문 dict을 사용하지 않으면 lux2는 무엇인가?
```

```python
lux3 = dict([("health", 490), ("mana", 334), ("melee", 550), ("armor", 18.72)])

print(lux3)
print(type(lux3))
```
