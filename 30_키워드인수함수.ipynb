```python
# 위치 인수를 사용하는 함수를 만들고 호출하기
def print_number(a, b, c):
    print(a)
    print(b)
    print(c)
```

```python
print_number(10, 20, 30)
```

```python
# 언패킹 사용하기
# *리스트를 각 변수로 만듬 
print_number(*[10, 20, 30])

# def unpacking():


# x = unpacking


a = [1, 2, 3, 4]
b = {1 : "a", 2 : "b"}
print(*a)
print(*b)
```

```python
print_number(a)
```

```python
print_number(b)
```

```python
# 가변인수 만들기

## def 함수이름():
##     코드

def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg)



```

```python
print_numbers(10)
print_numbers(10,20,30,40)
```

```python
print(10, 20, 30, 40)
```

```python
x = [10]
print_numbers(x)
```

```python
y = [10, 20, 30, 40]
print_numbers(*
              y)
```

```python
def print_numbers(a, *args):
    print(a)
    print(args)
    
```

```python
print_numbers(1,10,20)
```

```python
# 키워드 인수 사용하기

def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30,"서울시 용산구")
```

```python
personal_info(name = "홍길동",address = "서울시 용산구",age =  30)
```

```python
x = {
    "name" : "홍길동", 
    "age" : 30,
    "address" : "서울시 용산구"
    }


personal_info(**x)
```

```python
x = {
    "A" : "홍길동", 
    "B" : 30,
    "C" : "서울시 용산구"
    }


personal_info(*x)
```

```python
# 키워드 인수를 사영하는 가변 인수 함수 만들기

# def 함수이름(**매개변수):
#     코드

def personal_info(**kwargs):
    print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, ": ", arg, sep = "")
```

```python
personal_info(name = "홍길동")
```

```python
personal_info(name = "홍길동",address = "서울시 용산구",age =  30)
```

```python
# 매개변수에 초깃값 저장하기

# def 함수이름(매개변수= 값):
#     코드

def personal_info(name, age, address="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30)
```

```python
personal_info("홍길동", 30, "서초구 용산구 이촌동")
```

```python
# 초깃값이 지정된 매개변수의 위치

def personal_info(name, address="비공개", age="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```
