```python
# 대각선으로 별 출력하기

for i in range(int(input())):
    for k in range(10,-1,-1):
        if k < i:
            print("*",end="")
        else:
            print(" ",end="")
    for j in range(0,10):
        if j > i:
            print(" ",end="")
        else:
            print("*",end="")
    
    print()
```
