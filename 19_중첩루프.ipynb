```python
import sys

# 별의 크기를 조절하는 변수 (홀수 권장, 11~15 정도가 예쁩니다)
n = 11  

# 1. 윗부분 피라미드 (별의 위쪽 뿔)
# 이 부분은 일반적인 피라미드 모양입니다.
for i in range(n):
    # 공백 출력
    for j in range(n - i - 1):
        sys.stdout.write(" ")
    # 별 출력 (홀수 개수로 증가)
    for k in range(2 * i + 1):
        sys.stdout.write("*")
    sys.stdout.write("\n") # 줄 바꿈

# 2. 아랫부분 역피라미드 (별의 아래쪽 뿔 4개)
# 이 부분이 윗부분 삼각형과 겹쳐지면서 '별' 모양을 완성합니다.
# 윗부분의 중간 부분부터 시작하여 좁아지는 역피라미드를 그립니다.
for i in range(n // 2 + 1):
    # 공백 출력
    for j in range(i + n // 2 + 1): # 공백을 윗부분에 맞춰 더 많이 줍니다.
        sys.stdout.write(" ")
    # 별 출력 (홀수 개수로 감소)
    for k in range(n - 2 * i - 2): # 별의 개수를 줄여나갑니다.
        sys.stdout.write("*")
    sys.stdout.write("\n") # 줄 바꿈
```

```python
import turtle

# 6각별을 그리기 위한 펜 설정
t = turtle.Turtle()
t.speed(3) # 그리는 속도 조절
t.color("blue") # 선 색상

# 6각별을 구성하는 두 개의 삼각형 길이
length = 150

# 첫 번째 정삼각형 그리기
for _ in range(3):
    t.forward(length)
    t.left(120)

# 두 번째 삼각형을 그리기 위해 위치 이동 (펜을 들고 이동)
t.penup()
t.goto(0, length / 2) # 첫 번째 삼각형의 중간 높이로 이동
t.setheading(0) # 방향 초기화
t.pendown()

# 두 번째 정삼각형 그리기
for _ in range(3):
    t.forward(length)
    t.left(120)

# 그리기 완료 후 화면 유지
turtle.done()
```

```python
def draw_six_pointed_star(n):
    # n은 삼각형의 높이 (7~10 정도가 적당합니다)
    for i in range(-8, 8+1):
        for j in range(-2 * 8, 2 * 8 + 1):
            # 1. 위를 향하는 삼각형 영역 조건
            # y축(i)이 일정 범위 내에 있고, 양 변의 기울기 안에 있는지 확인
            up_triangle = (i <= 8 // 2) and (i >= -8) and (abs(j) <= (i + 8))
            
            # 2. 아래를 향하는 삼각형 영역 조건
            down_triangle = (i >= -8 // 2) and (i <= 8) and (abs(j) <= (8 - i))
            
            # 두 조건 중 하나라도 만족하면 별(*) 출력
            if up_triangle or down_triangle:
                print("*", end="")
            else:
                print(" ", end="")
        print() # 줄 바꿈

# 함수 실행 (크기 8로 설정)
draw_six_pointed_star(8)
```
