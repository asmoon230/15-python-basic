```python
# break 로 반복문 끝내기

i = 0
while True:
    print(i)
    i += 1
    if i == 10:
        break

```

```python
# for 에서 break로 반복문 끝낵

for i in range(10):
    print(i)
    if i == 5:
        break
```

```python
# continue로 코드 실행 건너뛰기

for i in range(10):
    if i % 2 == 0:
        continue #True 일시 넘어가라
    print(i)
```

```python
#while 반복문에서 코드 실행 건너뛰기

i = 0

while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 입력한 횟수 대로 반복하기

count = int(input("숫자"))

i = 0 
while True:
    print(i)
    i += 1
    if i == count:
        break


```

```python
# 입력한 숫자까지 홀수 출력하기

count = int(input("숫자"))

for i in range(count + 1):
    if i % 2 == 0:
        continue #반복문으로 돌아가라
    print(i)

```

```python
frist = int(input("1~200"))
second = int(input("10~200 단 첫번째 숫자 보다 크게"))

t = "3"
while True:
    s = str(frist)
    if t in s:
        frist += 1
        continue
    print(frist)
    if frist == second:
        break
    frist += 1




```

```python
frist = int(input("1~200 사이 첫 번째 숫자: "))
second = int(input("10~200 사이 두 번째 숫자: "))
t = "3"

while True:
    s = str(frist)
    
    if t in s:
        # 3이 포함되면 숫자를 하나 키우고 다음 반복으로 점프!
        frist += 1 
        continue 
    
    # 3이 포함되지 않은 경우에만 출력됨
    print(frist, "인식 결과: False (3 없음)")

    if frist == second:
        break
        
    frist += 1
```
