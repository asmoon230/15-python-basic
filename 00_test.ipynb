```python
# 문제 1

# students = {
#     "김철수": [90, 80, 70],
#     "이영희": [100, 95, 90],
#     "박민수": [60, 70, 65]
# }
# 출력결과 
# ===== 학생 성적 결과 =====

# 이름: 김철수

# 점수: [90, 80, 70]

# 평균: 80.0



# 이름: 이영희

# 점수: [100, 95, 90]

# 평균: 95.0



# 이름: 박민수

# 점수: [60, 70, 65]

# 평균: 65.0

# ===== 전체 결과 =====

# 전체 학생 수: 3명

# 전체 평균: 80.0
```

```python
students = {
    "김철수": [90, 80, 70],
    "이영희": [100, 95, 90],
    "박민수": [60, 70, 65]
}
s = list(students)
print("===== 학생 성적 결과 =====")

for i in s:
    print()
    print("이름 :",i)
    print()
    print("점수 :",students[i])
    k, e, m = students[i]
    t=int(k+e+m)/3
    print()
    print("평균 :",t)
    print()
print("===== 전체 결과 =====")
print()
print("전체 학생 수: ",len(students),"명")
print()
kcs1, kcs2, kcs3 = students["김철수"]
lyh1, lyh2, lyh3 = students["이영희"]
pms1, pms2, pms3,= students["박민수"]
allp = kcs1 + kcs2 + kcs3 + lyh1 + lyh2 + lyh3 + pms1 + pms2 + pms3
print("전체 평균: ",allp/9)
```

```python
# 문제 1. 
# 두 수 사이의 숫자 중  3과 7과 끝나지 않는 숫자 출력하기 

# 입력값 
# 3, 50

# 첫 번째 정수와 두 번째 정수 사이의 숫자 중 3 또는 7로 끝나지 않늣 숫자가 출력되게 만드세요 



# ---

a, b = list(map(int,input().split(",")))

b+=1

lis=[]
for k in range(a,b):
    if "3" in str(k) or "7" in str(k):
        continue
    lis.append(k)
print(*lis) 



```

```python
# 문제 2. 
# 표준 입력으로 역삼각형의 높이가 입력됩니다. 입력된 높이만큼 역삼각형의 "*"을 출력하는 프로그램을 작성하세요 

# 입력갑
# 5

# 출력값 
# *********
#  *******
#   *****
#    ***
#     *

# 요구사항:
# 중첩 for를 사용

# ---

p = int(input())

for j in range(p):        
    for i in range(p):
        if i > j:
            print("*", end="")
        else:
            print(" ", end="")
    for k in range(p,0,-1):
        if k > j:
            print("*", end="")
        else:
            print(" ", end="")
    print()

```

```python
# 문제 3. 
# 입력값을 참조하여 첫 번째 정수 부터 두 번째 정수까지 숫자를 역으로 출력하면서 
# 3의 배수일때는 Fizz, 7의 배수일때는 Buzz, 그외는 숫자를 출력하세요 

# 입력값 
# 100 10


# ---

a, b = list(map(int,input().split(" "))) 
a+=1
c=[]
c=list(range(b,a))
c.reverse()
for i in c:
    if i % 3 == 0 and i % 7 == 0:
        print("PizzBuzz")    
    elif i % 3 == 0:
        print("Pizz")
    elif i % 7 == 0:
        print("Buzz")
    else:
        print(i)



```

```python
# 문제 4
# 표준 입력으로 2개의 정수가 입력된다. 
# 첫번째 수와 두 번째 수 사이에 제곱근을 리스트(소수점 첫째자리)로 출력하는 프로그램을 만드세요 

# 입력값
# 1000 1200

# 요구사항
# 1. 리스트 컴프리헨션 표현식과 일반표현식으로 작성
# (2가지 방법)


# 1000과 1200사이에 있는 수의 제곱근들을 리스트로 만들어라
```

```python
a,b = list(map(int,input().split(" "))) 
b+=1

lis = [round(i**0.5,1) for i in range(a,b)]

print(lis)

lit=[]
for i in range(a,b):
    lit.append(round(i**0.5,1))
print(lit)
```
