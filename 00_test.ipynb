```python
# 문제 2. 
# 표준 입력으로 역삼각형의 높이가 입력됩니다. 입력된 높이만큼 역삼각형의 "*"을 출력하는 프로그램을 작성하세요 

# 입력갑
# 5

# 출력값 
# *********
#  *******
#   *****
#    ***
#     *

# 요구사항:
# 중첩 for를 사용

# ---

p = int(input())

for j in range(p):        
    for i in range(p):
        if i > j:
            print("*", end="")
        else:
            print(" ", end="")
    for k in range(p,0,-1):
        if k > j:
            print("*", end="")
        else:
            print(" ", end="")
    print()

```

```python
# # 지뢰찾기 
# 문제 1. 
# 표준 입력으로 2차원 리스트의 가로(col)와 세로(row)가 입력되고 그 다음 줄부터 리스트의 요소로 들아갈 문자가 입력됩니다. 
# 이때 2차원 리스트 안세어 *는 지뢰이고, .은 지뢰가 압니다. 지뢰가 아닌 요소에는 인접한 지뢰의 개수를 출력하는 프로그램을 만드세요 

# matrix = [] 
# for i in range(row): 
#     matrix.append(list(input())) 


# 입력 
# 3 3
# .**
# *..
# .*.

# 출력
# 2**
# *43
# 2*1

col, row = map(int,input().split())

matrix = [] 
for i in range(row): 
    matrix.append(list(input())) 




,
```

```python
for i in range(row):
    for j in range(col):
        if matrix[i][j] == "*":
            print("*",end= "")
        else:
            count = 0
            for a in range(i - 1, i +2 ):
                for b in range(j - 1, j + 2):
                    if 0 <= a < row and 0 <= b < col:
                        if matrix[a][b] == "*":
                            count += 1

            print(count,end="")
    print()
```

```python
문제. 2

prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}

함수를 호출하면
order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 샌드위치=2)

실행결과
아메리카노 2개: 5000원
카페라떼 1개: 3500원
총 결제 금액: 8500원
--------------------
바닐라라떼 2개: 8000원
초코케이크 1개: 5500원
샌드위치 1개: 6000원
총 결제 금액: 19500원
--------------------
아메리카노 1개: 2500원
등록되지 않은 상품입니다: 허니브레드
총 결제 금액: 2500원
--------------------

요구사항 
1. 함수 이름은 order로 만드세요.
2. 함수는 **kwargs를 사용해야 합니다.
3. kwargs의 key는 상품명, value는 주문 수량입니다.
```

```python
prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}


def order(**kwargs):
    lis=[]
    
    for key, value in kwargs.items():
        if key in prices:
            price=prices[key]
            price_= value * price
            print(f"{key} {value}개 : {price_:,}원")
            lis.append(price_)
            prices_total = sum(lis)
        else:    
            print(f"등록되지 않은 상품입니다: {key}")
    print(f"총 결제 금액: {prices_total:,}원")
    print("-"*20)


order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 허니브레드=2)
```

```python
# 문제 1. 재귀 호출로 합계와 계산 과정 출력하기

# 정수 n을 입력받아 n부터 1까지 더한 결과를 출력하는 프로그램을 만드세요.

# 단, 반드시 재귀 호출을 사용해야 하며, 계산 과정도 함께 출력해야 합니다.

# 입력:
# 5

# 출력: 
# 5 + 4 + 3 + 2 + 1 = 15

# 요구사항: 
# 1. 함수 이름은 recursive_sum으로 만드세요.
# 2. 반복문 for, while을 사용하지 마세요.
# 3. 재귀 호출을 사용하세요.
# 4. n == 1일 때 재귀 호출이 끝나도록 하세요.
# 5. 함수는 계산 과정 문자열과 합계를 함께 반환해야 합니다.
# 6. 최종 출력은 계산과정 = 결과 형태로 출력하세요.


# ---

def recursive_sum(n):
    if n == 1:
        print(n,"=")
        return 1 

    print(n,"+", end=" ")

    return n + recursive_sum(n-1)

print(recursive_sum(int(input())))
```
