```python
# for 와 range 사용하기

# for 변수 in range(횟수):
#     반복할 코드

# in 말고 다른 걸 사용 가능?

# range 말고 다른 함수 사용 가능?

for _ in range(100):    #100 번을 해라
    print("hello world")
```

```python
# 반복문에서 변수의 변화 알아보기

for i in range(10):
    print("hello world", i) #i 는 지역변수가 아니다.
print(i)
```
