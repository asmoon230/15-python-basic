```python
# 덧셈 함수만들기
## def 함수이름(매개변수1, 매개변수2):
##    코드
def add(a,b):
    print(a + b)
```

```python
y = add(10,20)
print(y)
```

```python
# 함수의 결과를 반환하기

def add(a, b):
    return a+b
x = add(10,20)
print(x)
```

```python
# 함수에서 값을 여러개 반환하기

def add_sub(a,b):
     
    return a + b, a - b 
x = add_sub(10,20)

print(x)
```

```python
# 함수의 호출과정 보기

def mul(a, b):
    c = a * b
    return c

def add(a, b):
    c = a + b
    print(c)
    d = mul (a ,b)
    print(d)

x = 10
y = 20

add(x,y)
```

```python
x, y = map(int,input().split())

def calc(a,b):
    c = a + b
    d = a - b
    e = a * b
    z = a / b
    f = float(z)
    return c, d, e, f

a , s, m, d = calc(x, y)

print('덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}'.format(a, s, m, d))

```
