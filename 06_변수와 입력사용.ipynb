```python
# 변수 여러 개를 한번에 만들기

x , y, z = 10, 20, 30
print(x)
print(y)
print(z)
```

```python
x, y = 10, 20
x, y = y, x

print(x)
print(y)
```

```python
# "hello python".split()
a, b = ['hello', 'python', "text"] #-> 리스트일때는 갯수가 서로 맞아야 한다

print(a)
print(b)
```

```python
# 두 숫자의 합 구하기

a, b = input("숫자 2 개를 입력하세요: ").split()
print(a)
print(b)
print(a + b)
```

```python
# 두 숫자의 합 구하기

a, b = input("숫자 2 개를 입력하세요: ").split()

a = int(a)
b = int(b)

print(a + b)
```

```python
# 입력받은 값을 콤마를 기준으로 분리하기
a, b = map(int, input("숫자 두 개를 입력하세요: ").split())

# a, b = map(int, ["10", "20"])
# 각 값에 지정한 함수를 적용하라
# a, b = 10, 20

print(a + b)
```

```python
a, b = map(int, input("숫자 두 개를 입력하세요: ").split(","))

print(a + b)
```
