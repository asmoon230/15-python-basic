```python
# 1부터 100까지 숫자 출력하기

for i in range(10):
    print(i)
```

```python
# 3 의 배수 fizz, 5의 배수 buzz

for i in range(10):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 ==0:
        print("buzz")
    else:
        print(i)
```

```python
for i in range(10):
    if i % 3 == 0 and i % 5 == 0:
        print("Fizz Buzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 ==0:
        print("buzz")
    else:
        print(i)
```

```python
#논리연산자를 사용하지 않고 3과 5의 공배수 처리하기
for i in range(20):
    if i % 15 == 0 :
        print("Fizz Buzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 ==0:
        print("buzz")
    else:
        print(i)
```

```python
#코드 단축하기

for i in range(21):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or i)
```
